```python
resume_selected = resume_df[
    [
        "Skills",
        "Experience (Years)",
        "Education",
        "Certifications",
        "Projects Count",
        "Job Role" 
    ]
]

resume_selected.head(20)
```

```python
jobs_selected = jobs_df[
    [
        "Title",
        "Company",
        "Industry",
        "Job.Description",
        "Employment.Type",
        "Education.Required",
    ]
]

jobs_selected.head()
```

```python
resume_clean = resume_selected.copy()
jobs_clean = jobs_selected.copy()
```

```python
resume_clean["Certifications"] = resume_clean["Certifications"].fillna("")

jobs_clean["Company"] = jobs_clean["Company"].fillna("Unknown")
jobs_clean["Industry"] = jobs_clean["Industry"].fillna("Not Specified")
jobs_clean["Job.Description"] = jobs_clean["Job.Description"].fillna("")
jobs_clean["Education.Required"] = jobs_clean["Education.Required"].fillna("Not Specified")
jobs_clean["Employment.Type"] = jobs_clean["Employment.Type"].fillna("Not Specified")
```

```python
print("Resume missing values:")
print(resume_clean.isnull().sum())

print("\nJobs missing values:")
print(jobs_clean.isnull().sum())
```

```python
import re 

def clean_text(text):
    text = str(text) #Convert anything to a string
    text = text.lower() #Convert to lowercase
    text = re.sub(r"[^a[a-z0-9\s]", " ", text) #Remove punctuation and special characters
    text = re.sub(r"\s+", " ", text).strip() #Remove extra spaces

    return text
```

```python
print("Original:")
print(resume_clean.loc[0, "Skills"])

print("\nCleaned:")
print(clean_text(resume_clean.loc[0, "Skills"]))
```

```python
print("Original:")
print(jobs_clean.loc[0, "Job.Description"][:300])

print("\nCleaned:")
print(clean_text(jobs_clean.loc[0, "Job.Description"])[:300])
```

```python
resume_clean["Resume_Text"] = (
    resume_clean["Skills"] + " " + 
    resume_clean["Experience (Years)"].astype(str) + " years " +
    resume_clean["Education"] + " " + 
    resume_clean["Certifications"] + " " +
    resume_clean["Projects Count"].astype(str) + " projects"
)
```

```python
jobs_clean["Job_Text"] = (
    jobs_clean["Title"] + " " +
    jobs_clean["Company"] + " " +
    jobs_clean["Job.Description"] + " " 
)
```

```python
resume_clean["Resume_Text"] = resume_clean["Resume_Text"].apply(clean_text)

jobs_clean["Job_Text"] = jobs_clean["Job_Text"].apply(clean_text)
```

```python
print("=== Resume Text ===")
print(resume_clean.loc[0, "Resume_Text"])

print("\n=== Job Text ===")
print(jobs_clean.loc[0, "Job_Text"])
```
